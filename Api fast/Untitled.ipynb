```python
import os

# Chemin du dossier contenant les données
dataset_path = "dataset mixte"

# Initialisation du dictionnaire pour stocker les résultats
image_counts = {}

# Parcourir chaque dossier de niveau supérieur (sain, légère, sévère)
for class_folder in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_folder)
    if os.path.isdir(class_path):  # Vérifier si c'est un dossier
        image_counts[class_folder] = {}
        
        # Parcourir les sous-dossiers (nutriments)
        for nutrient_folder in os.listdir(class_path):
            nutrient_path = os.path.join(class_path, nutrient_folder)
            if os.path.isdir(nutrient_path):  # Vérifier si c'est un sous-dossier
                # Compter les fichiers dans le sous-dossier
                num_images = len([file for file in os.listdir(nutrient_path) 
                                  if os.path.isfile(os.path.join(nutrient_path, file))])
                image_counts[class_folder][nutrient_folder] = num_images

# Afficher les résultats
for class_folder, nutrient_counts in image_counts.items():
    print(f"Classe '{class_folder}':")
    for nutrient_folder, count in nutrient_counts.items():
        print(f"  - {nutrient_folder}: {count} images")
```
